```python
import matplotlib.pyplot as plt #help create a customize chart 
import seaborn as sns #helps in data visualization

plt.figure(figsize=(14, 7)) #specifies thee dimensions of the chart

sns.countplot(x="G3", data=df)#  we want to use the student dataset called df,examine the G3 column, and create a bar chart showing how many students achieved each grade.

plt.title("Distribution of Students' Final Grades")
plt.xlabel("Final Grade (G3)")
plt.ylabel("Number of Students")

plt.show()
```

```python
plt.figure(figsize=(10, 5))

sns.countplot(x="studytime", data=df)

plt.title("Distribution of Students' Study Time")
plt.xlabel("Study Time")
plt.ylabel("Number of Students")

plt.show()
```
